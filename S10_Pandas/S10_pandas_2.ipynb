```python
from datetime import datetime

# Función para preguntar la fecha manualmente, aunque no es necesario porque coge la información del dataframe
def preguntar_fecha():
    while True:
        try:
            dia = int(input("Introduce el día en el que naciste: "))
            mes = int(input("Introduce tu mes de nacimiento: "))
            año = int(input("Introduce el año de nacimiento: "))
            break
        except ValueError:
            print("Introduce solo números enteros.")
    
    fecha_nacimiento = datetime(year=año, month=mes, day=dia)
    return fecha_nacimiento

# Función para calcular la edad
def calcular_edad(fecha_nacimiento):
    hoy = datetime.today()
    return (hoy - fecha_nacimiento).days // 365

# Llamada de las funciones
fecha = preguntar_fecha()
calcular_edad(fecha)
```

1.1 Crea una funció que prengui un dataframe com a paràmetre d'entrada.
La funció ha de crear (i exportar) un gràfic automàticament per a cada columna del dataframe.
Per exemple:
- histograma/boxplot si la variable és numèrica
- barres dels valors més freqüents si és categòrica
- barres dels anys més freqüents si la dada està en format data.

La idea és crear una funció que funcioni per qualsevol dataframe, no només amb el que hem treballat fins ara.
Mostra el resultat de la funció en algun dels datasets d’exemple que conté el paquet seaborn.



```python
import matplotlib.pyplot as plt
import pandas as pd

def crear_grafico(df):
    for columna in df.columns:
        plt.figure()

        if pd.api.types.is_numeric_dtype(df[columna]):
            df[columna].hist()

        elif pd.api.types.is_datetime64_any_dtype(df[columna]):
            df[columna].dt.year.value_counts().sort_index().plot(kind = "bar")
        
        else:
            df[columna].value_counts().head(10).plot(kind = "bar")

        plt.title(columna)
        plt.tight_layout()
        plt.savefig(f"{columna}.png")
        plt.show()
        plt.close()
```

```python
import seaborn as sns

df_pen = sns.load_dataset("penguins")
df_pen.head(10)
```

```python
crear_grafico(df_pen)
```

2.3 Ens interessa visitar totes les ciutats principals d'Espanya recorrent la mínima distància possible. Desenvolupa una solució raonable utilitzant les eines que tens actualment.
Per exemple, una aproximació senzilla (que no òptima) seria anant sempre a la ciutat més propera que no haguem visitat encara. 

- Fes una funció que donada la matriu de distàncies i la ciutat d'origen, faci una proposta de ruta que sigui el més curta possible que puguis.
- Retorna una llista amb l'ordre de visita.
- Dóna també la distància total recorreguda.


```python
def crear_ruta(matriz, inicio):

    no_visitadas = list(matriz.index)
    no_visitadas.remove(inicio)

    viaje = [{"Ciudad": inicio, "Km desde anterior": 0}]
    ciudad_actual = inicio
    km_realizados = 0

    while no_visitadas:
        distancia = matriz.loc[ciudad_actual, no_visitadas]
        proxima = distancia.idxmin()
        km_etapa = distancia[proxima]
        km_realizados += km_etapa

        viaje.append({
            "Ciudad": proxima,
            "Km desde anterior": km_etapa,})

        no_visitadas.remove(proxima)
        ciudad_actual = proxima

    df_viaje = pd.DataFrame(viaje)
    
    
    total = pd.DataFrame([{"Ciudad": "TOTAL", "Km desde anterior": round(km_realizados, 2)}])
    df_viaje = pd.concat([df_viaje, total], ignore_index=True)

    return df_viaje
```

```python
crear_ruta(df_rutas, "Barcelona")
```

(EXTRA) Des de quina ciutat la ruta seria més curta amb l'algoritme plantejat 

```python
def ruta_corta(matriz):
    dict_ruta = {}
    
    for ciudad in matriz.index:
        df_ruta = crear_ruta(matriz, ciudad)
        km_total = df_ruta.loc[df_ruta["Ciudad"] == "TOTAL", "Km desde anterior"].values[0] # lo mismo que km_total = df_ruta.iloc[-1, -1], última fila, última columna
        dict_ruta[ciudad] = km_total

    ciudad_min = min(dict_ruta, key=dict_ruta.get)

    return f"La ruta más corta comenzaría en: {ciudad_min} y sería de {dict_ruta[ciudad_min]} kilómetros"

ruta_corta(df_rutas)
```
