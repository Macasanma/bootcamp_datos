Consell: Intenta validar les dades prèviament, perquè enviï un missatge d’advertència si les dades introduïdes per l’usuari/a no estan en el format adequat o no pren valors raonables.


#### Función para solicitar los datos al usuario:

```python

def pedir_datos(informacion,restriccion):
    """
    Pide al usuario que introduzca la información, comprueba que este en el formato adecuado y con valores razonables.
    Devuelve el dat en float y con notación decimal anglosajona.
    """

    while True:                                        
        try:                                            
            dato = float(input(informacion).replace(",",".")) 
            if dato <= 0:
                print("Valor no válido")
            elif dato > restriccion:
                print(f"El valor no puede ser superior a {restriccion}")
            else:
                return dato                          
        
        except ValueError:                  
            print("Introduce la información con caracteres numéricos")
```

#### Función para realizar los cálculos del IMC:

```python

def calculo_IMC(weight, height):
    """
    Calcula el índice de masa corporal del usuario y lo clasifica en diferentes categorías según el resultado.
    """
    imc = weight / (height ** 2)
    
    if imc < 18.5:
        print(f"Su índice de masa corporal es {imc:.2f} e indica bajo peso.")
    elif imc < 25:
        print(f"Su índice de masa corporal es {imc:.2f} e indica peso normal.")
    elif imc < 30:
        print(f"Su índice de masa corporal es {imc:.2f} e indica sobrepeso.")
    else: 
        print(f"Su índice de masa corporal es {imc:.2f} e indica obesidad.")
    return
```

```python
# variables para ejecutar el cálculo de IMC con la restricción máxima:
peso = pedir_datos("Introduzca su peso en kilogramos: ", 200)
altura = pedir_datos("Introduzca su altura en metros: ", 2.10)

calculo_IMC(peso,altura)
```

###  Ejercicio 2: Convertidor de temperatures

Existeixen diverses unitats de temperatura utilitzades en diferents contextos i regions. Les més comunes són Celsius (°C), Fahrenheit (°F) i Kelvin (K). També existeixen altres unitats com Rankine (°Ra) i Réaumur (°Re).

Selecciona almenys 2 conversors, de tal manera que en introduir una temperatura retorni, com a mínim, dues conversions, de manera que es puguin guardar (recorda que un print() no es pot guardar mai).
Consell: Intenta validar les dades prèviament, perquè enviï un missatge d’advertència si les dades introduides per l’usuari/a no estan en el format adequat.

(EXTRA): Pensa una manera d’emmagatzemar totes les possibles conversions en un sol objecte (Llista? Diccionari? DataFrame?) en comptes d’escriure molts if else en funció de la temperatura d’origen i la temperatura de destí.

#### Función para solicitar los datos al usuario:

```python
def pedir_datos_temperatura():
    """
    Pide al usuario que introduzca la temperatura y la unidad comprueba que esten en el formato adecuado.
    Devuelve la temperatura y la unidad.
    """
    # Pedir temperatura y comprobar formato
    while True:                                        
        try:                                            
            temp = float(input("Introduce la temperatura: ").replace(",", "."))                                 
            break
        except ValueError:                  
            print("Introduce la información con caracteres numéricos")

    # Pedir unidad y comprobar formato 
    while True:
        unid = input("Introduce la unidad: Celsius (C), Fahrenheit (F) o Reaumur (RE): ").upper()

        if unid not in ["C", "F", "RE", "CELSIUS", "FAHRENHEIT", "REAUMUR"]:
            print("La unidad no es correcta. Usa C, F o RE")
            continue
        break
        
    return temp, unid
```

#### Función para realizar la conversión:

```python
def conversor_temperatura(temperatura, unidad):
    """
    Convierte la temperatura a unidades celsius, fahrenheit o reaumur.
    Devuelve un diccionario con las conversiones mostrando la temperatura y unidad correspondientes.
    """

    if unidad in ["C", "CELSIUS"]:
        c = temperatura
        f = temperatura * 9/5 + 32
        re = temperatura * 4/5
    
    elif unidad in ["F", "FAHRENHEIT"]:
        c = (temperatura - 32) * 5/9
        f = temperatura
        re = c * 4/5

    elif unidad in ["RE", "REAUMUR"]:
        c = temperatura * 5/4
        f = c * 9/5 + 32
        re = temperatura
    
    else:
        raise ValueError ("Unidad no válida")


    return {"Celsius": round(c, 2), "Fahrenheit": round(f, 2), "Reaumur": round(re, 2)}
```

#### Ejecución de las funciones y resultados:

```python

temperatura, unidad = pedir_datos_temperatura()
resultado = conversor_temperatura(temperatura, unidad)

print(f"Resultado de la conversión: {resultado}")
```

###  Ejercicio 3: Comptador de paraules d'un text.

Escriu una funció que donat un text, mostri les vegades que apareix cada paraula. Intenta que es gestionin totes les casuístiques possibles que facin que el programa no funcioni correctament.
(EXTRA): Quina és la llargada mitjana de les paraules del text que has escrit? “Hola com va?” hauria de retornar (4+3+2) / 3 = 3


#### Función para solicitar los datos al usuario:

```python
import re

def pedir_texto():
    """
    Solicita un texto al usuario.
    Normaliza el texto a minúsculas y extrae palabras eliminando los signos de puntuación y símbolos.
    Devuelve una lista de palabras (strings).
    """
    
    texto = input("Escribe tu texto aquí: ").lower()
    palabras = re.findall(r"\b\w+\b", texto)
    
    return palabras
```

#### Función para mostrar las veces que aparece una palabra:

```python
def contar_palabras(lista_palabras):
    """
    Recibe una lista de palabras (strings).
    Devuelve un diccionario con las palabras y el número de veces que aparecen.
    """
    dict_palabras = {}

    for palabra in lista_palabras:
        if palabra in dict_palabras:
            dict_palabras[palabra] += 1
        else:
            dict_palabras[palabra] = 1
    
    return dict_palabras
```

#### Función para calcular la longitud media de las palabras:

```python
def calcular_media_longitud(palabras):
    """
    Calcula la longitud media de las palabras de una lista.
    """
    contador_letras = 0

    for palabra in palabras:
        contador_letras += len(palabra)

    media = contador_letras / len(palabras)

    return media
```

#### Ejecución de la función y resultado:

```python
texto = pedir_texto()

diccionario = contar_palabras(texto)
media = calcular_media_longitud(texto)

print("Diccionario:", diccionario)
print("Media longitud:", round(media, 2))
```

###  Ejercicio 4:  Diccionari invers (amb possibilitat de duplicats)
Resulta que el client té una enquesta molt antiga que s'emmagatzema en un diccionari i els resultats els necessita al revés, és a dir, intercanviant les claus i els valors.

Els valors i claus en el diccionari original són únics; si aquest no és el cas, la funció hauria d'imprimir un missatge d'advertiment, juntament amb una llista amb els valors associats a la clau repetida.
 

#### Función para solicitar los datos a un usuario:

```python
def pedir_diccionario():
    """
    Solicita el número de parejas para crear un diccionario.
    Devuelve un diccionario.
    """
    numero_parejas = int(input("¿De cuántas parejas de clave-valor quieres que se componga el diccionario?"))
    dict_principal = {}
    
    for i in range(numero_parejas):
        clave = input("Introduce una clave: ")
        valor = input("Introduce un valor: ")
        dict_principal[clave] = valor
    
    return dict_principal
```

#### Función para invertir un diccionario y detectar duplicados:

```python
def invertir_diccionario(dic):
    """
    Recibe un diccionario, invierte las claves y los valores.
    Detecta duplicados y los guarda en otro diccionario.
    Devuelve una tupla con el diccionario invertido y el duplicado.
    """
    dict_inverso = {}
    dict_duplicado = {}

    for clave, valor in dic.items():
        if valor not in dict_inverso:
            dict_inverso[valor] = clave

        else:
            print(f"Aviso: el valor {valor} está duplicado.")
            dict_inverso[valor] = [dict_inverso[valor], clave]
            dict_duplicado[valor] = dict_inverso[valor]

    return dict_inverso, dict_duplicado
```

#### Ejecución de la función y resultado:

```python
diccionario = pedir_diccionario()
invertido, duplicado = invertir_diccionario(diccionario)

print(f"El diccionario inverso es: {invertido}")
print(f"El diccionario de duplicados es: {duplicado}")
```

###  Ejercicio 1 : Comptador i endreçador de paraules d'un text.

El client va quedar content amb el comptador de paraules, però ara vol llegir arxius TXT i que calculi la freqüència de cada paraula ordenades dins de les entrades habituals del diccionari segons la lletra amb la qual comencen, és a dir, les claus han d'anar de la A a la Z i dins de la A hem d'anar de la A la Z.

#### Función para importar un texto, normalizarlo y pasarlo a string:

```python
from pathlib import Path

def cargar_normalizar_texto(nombre_archivo):
    """
    Lee un archivo de texto y lo normaliza convirtiéndolo en minúsculas y eliminando signos de puntuación o caracteres.
    Devuelve el texto en formato lista separado por palabras.
    """
    texto = Path(nombre_archivo).read_text(encoding="utf-8").lower()
    palabras = re.findall(r"\b\w+\b", texto)

    return palabras
```

#### Función para crear un diccionario con las palabras del texto como claves y el valor con el número de veces que aparecen.

```python
def contar_agrupar_dict(lista_texto):
    """
    Recibe una lista de palabras y las agrupa en un diccionario las palabras por su primera letra.
    Cada letra contiene las palabras que empiezan por ella y cuántas veces aparecen.
    Devuelve un diccionario ordenado alfabéticamente.
    """

    dict_abecedario = {}

    for palabra in lista_texto:
        # Agrupa según la primera letra
        
        letra = palabra[0]
        if letra not in dict_abecedario:
            dict_abecedario[letra] = {}
        
        # Cuenta las veces que aparece la palabra en el texto
        if palabra in dict_abecedario[letra]:       
            dict_abecedario[letra][palabra] += 1
        else:
            dict_abecedario[letra][palabra] = 1
    
    dict_final = {}
        
    for letra in sorted(dict_abecedario.keys()):
            palabras = dict_abecedario[letra]
            palabras_orden = dict(sorted(palabras.items()))

            dict_final[letra] = palabras_orden

    return dict_final
```

#### Ejecución de las funciones y resultado:

```python
texto = cargar_normalizar_texto("tu me quieres blanca.txt")
resultado = contar_agrupar_dict(texto)

for letra in resultado:
    print(f"\n{letra.upper()}")
    print(resultado[letra])
   
```

#### Ejercicio 2: Conversió de tipus de dades.
El client rep una llista de dades i necessita generar dues llistes, la primera on estaran tots els elements que es van poder convertir en flotants i l'altra on estan els elements que no es van poder convertir. 
Exemple de la llista que rep el client:
[ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']]


```python
def convertir_float(lista_elementos):
    """
    Recorre una lista que recibe (incluyendo, si tiene listas y tuplas internas)  
    Devuelve dos listas: los que se pueden convertir a float y los que no.
    """
    lista_floats = []
    lista_otros = []

    for elemento in lista_elementos:

        if isinstance(elemento, (list, tuple)):
            for subelemento in elemento:
                try:
                    lista_floats.append(float(subelemento))
                except:
                    lista_otros.append(subelemento)
        else:
            try:
                lista_floats.append(float(elemento))
            except:
                lista_otros.append(elemento)

    return lista_floats, lista_otros
```

#### Ejecutar la función y mostrar los resultados:

```python
lista_datos = ([ '1.3', 'one' , '1e10' , 'seven', '3-1/2', ('2',1,1.4,'not-a-number'), [1,2,'3','3.4']])
floats, otros = convertir_float(lista_datos)

print(f"La lista con elementos transformados a float es: {floats}")
print(f"La lista con elementos que no se pueden transformar a float es: {otros}")
```

#### Ejercicio 1. Generador de Contrasenyes

Explora el funcionament del mòdul random de la llibreria numpy. 🔗 Random module in NumPy
El client ha detectat un problema amb les contrasenyes que utilitzen els seus treballadors. Asdf1234, dates d’aniversari o similars. Per a solucionar-ho ens ha encarregat una funció de Python que generi contrasenyes més segures. La funció ha de dependre dels següents paràmetres:

longitud (int): Longitud de la contrasenya
majuscules (bool = True): Si hi ha d’aparèixer majúscules
minuscules (bool = True): Si hi ha d’aparèixer minúscules
numeros (bool = True): Si hi ha d’aparèixer números
signes (bool = False): Si hi ha d’aparèixer caràcters especials (,-$? o similars)

Així doncs, si executem la funció de la següent manera:
crear_contrasenya(10, True, True, True, True)
Hauríem d’obtenir un output (que hem de poder guardar) de l’estil:
9Er,5Vn8P$
Assegura’t que es compleixin tots els criteris, i que aquestes contrasenyes siguin realment aleatòries.

(EXTRA) Explora com podríem fer que la funció copiés la contrasenya automàticament al porta papers de l’ordinador (com si l’haguéssim seleccionat i fet ctrl+copy).

```python
import numpy as np
import random
import string

def generador_contraseñas(longitud, mayusculas=True, minusculas=True, numero=True, signos=False):

    mayus = np.array(list(string.ascii_uppercase))
    minus = np.array(list(string.ascii_lowercase))
    num = np.array(list(string.digits))
    sign = np.array(list("!@#$%&?,.-¿¡+*;:"))

    contraseña = []

    if mayusculas:
        contraseña.append(np.random.choice(mayus))

    if minusculas:
        contraseña.append(np.random.choice(minus))

    if numero:
        contraseña.append(np.random.choice(num))

    if signos:
        contraseña.append(np.random.choice(sign))

    if longitud < len(contraseña):
        raise ValueError("Longitud insuficiente")

    faltan = longitud - len(contraseña)

    if faltan > 0:
        todo = np.concatenate((mayus, minus, num, sign))
        contraseña.extend(np.random.choice(todo, faltan))

    random.shuffle(contraseña)

    return "".join(contraseña)
```

#### Ejecutar la función y mostrar el resultado:

```python
contraseña = generador_contraseñas(8, True, True, True, True)

import pyperclip
pyperclip.copy(contraseña)

print("Tu nueva contraseña es: ", contraseña)
```

#### Ejercicio 2. Processament de dades simple:

Té un històric de partits de futbol català en un fitxer, on hi ha emmagatzemat els noms dels equips i els resultats. Necessita que processem les dades de manera automàtica, per extreure’n els resultats que necessita.
Utilitza el fitxer "historic_partits.txt"
Necessita un programa que retorni:

El nombre total de gols que ha fet cada equip.
El nom de l’equip més golejador.
El nom de l’equip més golejat
La classificació global (cada victòria: 3 pts, empat 1 pts, derrota 0 pts)



#### Función para cargar un archivo de texto:

```python
from pathlib import Path

def cargar_texto(nombre_archivo):
    """
    Lee un archivo de texto.
    Devuelve una lista de cadenas.
    """
    texto = Path(nombre_archivo).read_text(encoding="utf-8").lower()
    return texto.splitlines()
```

#### Función para organizar la información de los partidos:

```python
def categorizar_partidos(archivo_partidos):
    """
    Recorre los partidos y calcula por equipo goles marcados o recibidos.
    Devuelve dos diccionarios con los goles a favor o en contra de cada equipo.
    """

    goles_ganados = {}
    goles_perdidos = {}

    for elemento in archivo_partidos:
        juego = elemento.strip().split("\t")

        equipo_local = juego[0]
        resultado = juego[1]
        equipo_visitante = juego[2]

        goles1 = int(resultado.split("-")[0])
        goles2 = int(resultado.split("-")[1])

        
        for equipo in [equipo_local, equipo_visitante]:
            if equipo not in goles_ganados:
                goles_ganados[equipo] = 0
                goles_perdidos[equipo] = 0

        # sumar los goles según el equipo
        goles_ganados[equipo_local] += goles1
        goles_ganados[equipo_visitante] += goles2

        goles_perdidos[equipo_local] += goles2
        goles_perdidos[equipo_visitante] += goles1

    return goles_ganados, goles_perdidos
```

#### Función para calcular puntos de los partidos:

```python
def calcular_puntuacion(partidos):
    """
    Calcula la puntuación según los partidos y goles.
    Gana = 3, empata = 1, pierde = 0
    """
    puntos = {}

    for elemento in partidos:
        equipo1,resultado,equipo2 =  elemento.strip().split("\t")

        goles1 = int(resultado.split("-")[0])
        goles2 = int(resultado.split("-")[1])

        for equipo in(equipo1, equipo2):
            puntos.setdefault(equipo,0)

        puntos[equipo1] += 3 if goles1 > goles2 else 1 if goles1 == goles2 else 0
        puntos[equipo2] += 3 if goles2 > goles1 else 1 if goles1 == goles2 else 0
     
    return puntos
```

#### Función para calcular los resultados finales del partido:

```python
def resultados_finales(goles_win, goles_lost, puntos):
    """
    Calcula los principales resultados del torneo:
    máximo goleador, más goleado, ranking de goles y clasificación.
    """
    goleador_estrella = max(goles_win, key=goles_win.get)
    perdedor_estrella = max(goles_lost, key=goles_lost.get)
 
    goles_ordenados = sorted(goles_win.items(), key=lambda x: x[1],reverse=True)
    clasificacion = sorted(puntos.items(), key=lambda x: x[1],reverse=True)

    return {
        "max_goleador": goleador_estrella,
        "max_goleado": perdedor_estrella,
        "goles_ranking": goles_ordenados,
        "ranking": clasificacion
        }
```

```python
partidos = cargar_texto("historic partits.txt")
goles_win, goles_lost = categorizar_partidos(partidos)
puntos = calcular_puntuacion(partidos)
resultados = resultados_finales(goles_win, goles_lost, puntos)

print("MÁXIMO GOLEADOR:", resultados["max_goleador"])
print("MÁXIMO GOLEADO:", resultados["max_goleado"])

print("\nGOLES POR EQUIPO (ordenados):")
for equipo, goles in resultados["goles_ranking"]:
    print(equipo, goles)

print("\nCLASIFICACION:")
for equipo, puntos in resultados["ranking"]:
    print(equipo, puntos)
```
